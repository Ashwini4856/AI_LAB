```python

print("Nandini kakade")
print("1BM25CS487")

def is_variable(x):
    return isinstance(x, str) and x[0].islower()


def substitute(x, subst):
    while isinstance(x, str) and x in subst:
        x = subst[x]
    return x


def unify(x, y, subst):

    x = substitute(x, subst)
    y = substitute(y, subst)

    if x == y:
        return True

    if is_variable(x):
        subst[x] = y
        return True

    if is_variable(y):
        subst[y] = x
        return True

    return False


def parse_expression(expr):
    name, args = expr.split("(", 1)
    args = args.rstrip(")")
    return name.strip(), [a.strip() for a in args.split(",")]



A = input("Enter first expression: ")
B = input("Enter second expression: ")


p1, args1 = parse_expression(A)
p2, args2 = parse_expression(B)

subst = {}


if p1 != p2:
    print("Unification Failed: Predicate names are different")


elif len(args1) != len(args2):
    print("Unification Failed: Number of arguments are different")

else:
    success = True

    for x, y in zip(args1, args2):
        if not unify(x, y, subst):
            success = False
            break

    if success:
        print("\nUnification Successful!")
        print("Substitution:", subst)

        result1 = [substitute(x, subst) for x in args1]
        result2 = [substitute(x, subst) for x in args2]

        print("Expression A:", p1 + "(" + ", ".join(result1) + ")")
        print("Expression B:", p2 + "(" + ", ".join(result2) + ")")

        print("Both expressions are unified.")
    else:
        print("Unification Failed")
```
