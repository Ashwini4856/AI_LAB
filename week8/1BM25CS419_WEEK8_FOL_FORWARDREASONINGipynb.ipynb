```python
print("Ashwini P")
print("1BM25CS419")

def forward_chaining(facts, rules, query):

    while True:
        new = set()

        for premises, conclusion in rules:
            if all(p in facts for p in premises):

                if conclusion not in facts:
                    new.add(conclusion)

        if query in facts:
            return True

        if not new:
            return False

        facts.update(new)

n = int(input("Enter number of facts: "))
facts = set()

for i in range(n):
    fact = input("Enter fact: ")
    facts.add(fact)

r = int(input("\nEnter number of rules: "))
rules = []

for i in range(r):
    print("\nRule", i + 1)
    p = int(input("Enter number of premises: "))

    premises = []
    for j in range(p):
        premises.append(input("Enter premise: "))

    conclusion = input("Enter conclusion: ")

    rules.append((premises, conclusion))

query = input("\nEnter query to prove: ")


if forward_chaining(facts, rules, query):
    print("\nQuery proved: True")
else:
    print("\nQuery cannot be proved: False")
```
